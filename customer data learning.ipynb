```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#Load the dataset
df = pd.read_csv('customer_data_records.csv')
print(df.head(30))
```

```python
# Select features for clustering
X  = df[[ 'Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
#Perform KMeans clustering
kmeans = KMeans(n_clusters=5, random_state=42)
kmeans.fit(X)
```

```python
# Add cluster labels to the original DataFrame
df['Cluster'] = kmeans.labels_
print(df)
```

```python
#Visualize the clusters
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df , x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster', palette='viridis', s=100)
plt.title('Custemer clusters based on Annual Income and Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```
